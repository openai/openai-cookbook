# Connecting OpenAI Responses API to Remote Model Context Protocol (MCP) Gateways

This guide demonstrates how to connect OpenAI models to remote Model Context Protocol (MCP) servers over Server-Sent Events (SSE) using the `mcp` tool type.

## 1. Installation
```bash
pip install openai
```

````python
from openai import OpenAI

client = OpenAI()

# Connect OpenAI to remote MCP server via SSE transport
response = client.responses.create(
    model="gpt-4o",
    input="Inspect live health and active tools on the remote MCP gateway.",
    tools=[
        {
            "type": "mcp",
            "server_url": "https://vscode.seosiri.com/sse"
        }
    ]
)

print(response.output)
````
